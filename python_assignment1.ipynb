```python
#1. Second Largest Element
'''Write a Python program to find the second largest unique element in a list without using the
sort() function'''

numbers=[12,45,23,45,67,12]
largest=None
sec_largest=None
for i in numbers:
  if largest is None or i>largest:
    if i!=largest:
     sec_largest=largest
     largest=i
  elif i!=largest and (sec_largest is None or i>sec_largest):
    sec_largest=i
print("second largest unique element is: ",sec_largest)
```

```python
#2. Frequency Analysis
'''Write a program to find the frequency of each element in a list without using the count()
method'''

number=[1,4,6,2,4,1,2,8,6]
dict={}
for i in number:
  if i in dict:
    dict[i]+=1
  else:
    dict[i]=1
print("frequency",dict)
```
