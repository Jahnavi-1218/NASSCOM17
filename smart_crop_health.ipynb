```python
import numpy as np
import pandas as pd
df=pd.DataFrame({
    'day':[1,2,3,4,5,6],
    'soil moisture':[72,68,62,55,48,40],
    'Temperature':[27,29,31,33,35,37],
    'humidity':[78,74,70,65,60,55],
    'nitrogen':[48,46,43,39,35,30],
    'crop health':[94,90,84,76,65,52]
    })
df

```

```python
d=df.copy()
for i in d:
    print(d[i])
```

```python
d["status"]=np.where(
    d['soil moisture']<50,'low soil moisture alert','soil moisture normal'
)
print(d[['soil moisture','status']])
```

```python
d['condition']=np.where(
    d['Temperature']>35,'high temperature alert','normal temperature'
)
print(d[['Temperature','condition']])
```

```python
def crop_status(score):
  if score>80:
     return 'healthy'
  elif score>=50 and score<=80:
    return 'under stress'
  else:
    return 'critical'
d['status']=d['crop health'].apply(crop_status)
print(d[['crop health','status']])

```
